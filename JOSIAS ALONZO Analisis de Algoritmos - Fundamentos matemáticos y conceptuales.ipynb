**Techo** ($\lceil x \rceil$ o **ceiling**): Es el menor entero que es mayor o igual que $x$ (por ejemplo, $\lceil 6.1 \rceil = 7$).  
En el diseño y análisis de algoritmos, estas funciones se emplean en diversas áreas clave:

**2. Crecimiento de la Función Exponencial vs. Polinomial**
De acuerdo con los textos, una función exponencial positiva (de la forma $a^n$ con una base constante $a > 1$) siempre crece más rápido que cualquier función polinomial (de la forma $n^b$ donde el exponente $b$ es una constante real positiva)  
**El argumento matemático**  

Para demostrar esto de forma matemáticamente rigurosa, las fuentes recurren al cálculo de límites al infinito para analizar la tasa de crecimiento asintótico.  Se establece que para cualquier constante real $b$ y cualquier constante $a > 1$:$$\lim_{n \to \infty} \frac{n^b}{a^n} = 0$$A partir de este límite, se concluye formalmente que $n^b = o(a^n)$ (utilizando la notación "o pequeña"). Esto significa que, a medida que el tamaño de la entrada $n$ se vuelve suficientemente grande, el valor de la función polinomial se vuelve insignificante en comparación con el valor de la función exponencial. . Las potencias de $n$, por grandes que sean, siempre crecen de manera mucho más lenta que cualquier base exponencial positiva  
**Impacto en la práctica computacional**  
Esta diferencia matemática traza la frontera entre lo que es computable en la práctica y lo que no:Algoritmos polinomiales (Tratables): Aquellos cuya complejidad está acotada por un polinomio ($O(n^k)$) se consideran eficientes o tratables. Pueden resolver instancias grandes del problema en un tiempo razonable  
**Algoritmos exponenciales (Intratables)**: Aquellos cuya complejidad crece de forma exponencial ($O(a^n)$) se consideran ineficientes o intratables. Para entradas moderadamente grandes, el tiempo de ejecución requerido se dispara de forma absurda  
Por ejemplo, un algoritmo con tiempo de ejecución de $2^n$ ejecutándose en una computadora que realiza un billón ($10^{12}$) de operaciones por segundo tardaría aproximadamente $4 \cdot 10^{10}$ años en resolver una entrada de tamaño $n = 100$, un periodo de tiempo que supera por mucho la edad estimada del planeta Tierra

#1. Definición y Características Clave de un Algoritmo
Un algoritmo se define intuitivamente como una serie o secuencia finita de pasos para resolver un problema
. Formalmente, los textos lo caracterizan como un "conjunto de reglas operacionales inherentes a un cómputo", o bien como un conjunto finito de instrucciones precisas y detalladas que reciben unos valores de entrada, los procesan y los transforman en valores de salida. Para que un procedimiento sea formalmente considerado un algoritmo, debe poseer las siguientes características esenciales:    

**Entrada (Input)**: Debe tener definidas de forma clara las variables o datos iniciales que requiere para operar.

**Salida (Output)**: Debe especificar inequívocamente el resultado, solución o producto que genera al finalizar.  

**Precisión / No ambigüedad**: Cada instrucción y el orden exacto de los pasos deben estar explícitamente indicados de modo unívoco, de forma que no existan dudas sobre qué acción realizar en cada momento.   

**Determinismo**: Al ejecutarse múltiples veces utilizando el mismo conjunto de datos de entrada, el algoritmo debe producir exactamente el mismo resultado en cada ocasión.  

**Finitud**: Debe constar de un número limitado de pasos y garantizar que finalizará su ejecución tras concluir dicha secuencia. Si un método tiene los demás atributos pero no es finito, se le clasifica únicamente como un "método computacional".  

**Efectividad**: Cada acción planteada debe ser lo suficientemente básica para poder ejecutarse de forma factible en un tiempo finito.  

**Generalidad**: El procedimiento debe estar diseñado para contemplar y solucionar correctamente todas las posibles instancias o casos del problema planteado, y no solo un caso específico.  
#2. El Problema del Paro (Halting Problem) de Alan Turing  
Formulado en 1936 por el matemático británico Alan Turing mediante su célebre modelo matemático de la Máquina de Turing
, el Problema del Paro consiste en determinar de forma general si un programa de computadora arbitrario finalizará en algún momento su ejecución (se detendrá) o continuará ejecutándose indefinidamente (entrando en un bucle o ciclo infinito) al trabajar sobre una entrada dada.  

La aportación fundamental de Turing respecto a este problema establece que es asintóticamente imposible que exista un algoritmo o programa de computadora general capaz de resolver correctamente todas las instancias de este problema. Es decir, el Problema del Paro pertenece a la categoría de los problemas indecidibles o no computables.  

Los textos aclaran las limitaciones prácticas e intuitivas para intentar resolverlo:  
1. No es viable evaluar un código buscando patrones generales porque siempre habrá comportamientos y sutilezas lógicas complejas que escapen al análisis del patrón.  

2. Tampoco es una solución válida simplemente correr el programa en cuestión para ver si termina; si el programa se detiene sabremos que paró, pero si entra en un bucle infinito, la computadora continuará ejecutándolo por siempre y nunca podremos determinar en un tiempo finito si se detendría en el siguiente paso o si nunca lo hará. Esto invalida el intento como un algoritmo real porque no garantiza alcanzar una solución en pasos finitos  
#3 "Análisis A Priori" vs. "Prueba A Posteriori" para Medir la Eficiencia  
La evaluación de los recursos (tiempo de procesamiento y espacio en memoria) consumidos por un algoritmo consta de dos fases o metodologías complementarias:  
| Criterio de Comparación | Análisis A Priori (Teórico) | Prueba A Posteriori (Empírico / Experimental) |
| :--- | :--- | :--- |
| **Definición** | Consiste en obtener analítica y matemáticamente una función teórica que acote o limite el tiempo y la memoria del algoritmo en función del tamaño de sus datos de entrada ($n$). | Consiste en codificar el algoritmo en un programa real y medir directamente en la computadora los recursos consumidos (tiempo de reloj y bytes de memoria) durante su ejecución activa. |
| **Hardware y Software** | Es totalmente independiente del entorno físico. No requiere de una computadora, sistema operativo o compilador específico para ser calculado. | Es completamente dependiente del entorno. Las mediciones cambian según la velocidad del procesador, el compilador, la arquitectura física y la carga del sistema. |
| **Implementación** | Se realiza utilizando una representación abstracta o de alto nivel del algoritmo (como pseudocódigo) sin necesidad de programarlo. | Requiere obligatoriamente la labor de implementar, depurar y ejecutar el algoritmo en un lenguaje de programación específico (como Java o C). |
| **Entradas de datos analizadas** | Proporciona una predicción rigurosa que toma en consideración todas las posibles entradas y distribuciones de datos imaginables. | Se limita estrictamente a las mediciones de rendimiento obtenidas sobre un conjunto específico y limitado de ejemplares de prueba. |
| **Utilidad principal** | Permite predecir costos y comparar la eficiencia de distintos enfoques matemáticos desde la etapa de diseño de forma temprana, ahorrando horas de desarrollo laborioso. | Permite recolectar estadísticas reales y de laboratorio (mediante benchmarking) para comprobar el comportamiento práctico del programa frente a datos específicos del mundo real. |


- *Explica en qué consisten el peor caso, mejor caso y caso promedio al momento de evaluar el comportamiento de un algoritmo ante distintas entradas.*  
**RESPUESTA:**  
Al evaluar el rendimiento y consumo de recursos de un algoritmo, no basta con conocer únicamente el tamaño de la entrada ($n$). Como bien señalan los textos, entradas del mismo tamaño pueden hacer que un mismo algoritmo se comporte de maneras drásticamente diferentes según cómo estén distribuidos o estructurados los datos iniciales.

Para modelar matemáticamente esta variabilidad, el análisis de algoritmos define tres escenarios fundamentales: el peor caso, el mejor caso y el caso promedio.

### **1. El Peor Caso (Worst-Case)**

El peor caso representa el escenario más desfavorable posible para el algoritmo. Consiste en determinar la cota superior o el número máximo de operaciones básicas que el algoritmo ejecutará para cualquier entrada de tamaño $n$.

* **Definición matemática:** Si $D_n$ es el conjunto de todas las entradas posibles de tamaño $n$ y $t(I)$ es el tiempo o número de instrucciones ejecutadas con la entrada $I$, el peor caso $W(n)$ se define formalmente como:
  $$W(n) = \max \{ t(I) \mid I \in D_n \}$$
* **Importancia práctica:** Es el análisis más utilizado por los científicos de la computación. Ofrece una garantía absoluta de rendimiento; asegura que, bajo cualquier circunstancia, el tiempo de ejecución real del programa nunca superará ese límite. Además, en áreas críticas de software (sistemas de tiempo real, control aéreo, medicina), conocer el peor caso es crucial para evitar fallos catastróficos.

### **2. El Mejor Caso (Best-Case)**

El mejor caso es el escenario más optimista. Describe el comportamiento del algoritmo ante la entrada de tamaño $n$ que resulta ser la más favorable o conveniente, requiriendo el número mínimo de operaciones para finalizar.

* **Definición matemática:** Representa el mínimo del conjunto de operaciones posibles para entradas de tamaño $n$:
  $$B(n) = \min \{ t(I) \mid I \in D_n \}$$
* **Importancia práctica:** Aunque suele considerarse una medida "engañosa" porque rara vez ocurre en situaciones reales, el mejor caso no es inútil. Sirve como una herramienta de descarte rápido: si el mejor caso de un algoritmo ya es ineficiente o demasiado lento, podemos desechar el algoritmo de inmediato sin necesidad de hacer análisis más complejos. Asimismo, nos ayuda a identificar algoritmos que son extremadamente eficientes con datos que ya están "casi listos" (como el ordenamiento por inserción sobre listas casi ordenadas).

### **3. El Caso Promedio (Average-Case)**

El caso promedio describe el comportamiento "típico" u ordinario del algoritmo cuando se ejecuta repetidamente en el mundo real. En lugar de mirar los extremos, busca estimar la cantidad de trabajo esperada bajo una distribución de probabilidad de las entradas.

* **Definición matemática:** Se calcula multiplicando el costo de cada tipo de entrada por la probabilidad de que dicha entrada ocurra, sumando todos los valores para obtener la esperanza matemática:
  $$A(n) = \sum_{I \in D_n} P(I) \cdot t(I)$$
* **Importancia práctica:** Es la medida de rendimiento más realista y representativa para el software comercial o cotidiano. Muchos algoritmos con peores casos pésimos son ampliamente utilizados debido a que su caso promedio es excelente. El ejemplo clásico es Quicksort: su peor caso es un ineficiente $O(n^2)$, pero su caso promedio es de un óptimo $O(n \log n)$, lo que lo hace sumamente rápido y preferido en la práctica sobre otros métodos.
* **Dificultad de cálculo:** Es, por mucho, el análisis matemático más complejo. Requiere conocer o asumir de antemano la distribución probabilística de los datos de entrada (generalmente asumiendo que todas las entradas son equiprobables). Si esta suposición matemática es incorrecta, la estimación del caso promedio pierde su validez en la práctica.

### **Ejemplo Comparativo: Búsqueda Secuencial en un Arreglo**

Para ilustrar de forma intuitiva cómo difieren estos tres escenarios, consideremos el problema de buscar un número clave $K$ dentro de un arreglo desordenado de tamaño $n$:

1. **Mejor Caso:** Ocurre si la clave $K$ está ubicada en la primera posición del arreglo. El algoritmo hace exactamente $1$ comparación y termina (complejidad $O(1)$).
2. **Peor Caso:** Ocurre si la clave $K$ se encuentra en la última posición o, peor aún, no está presente en el arreglo. El algoritmo se ve obligado a recorrer todo el arreglo, realizando $n$ comparaciones antes de terminar (complejidad $O(n)$).
3. **Caso Promedio:** Si asumimos que la clave $K$ está en el arreglo y que tiene la misma probabilidad de estar en cualquier casilla, el número esperado de comparaciones será aproximadamente de $n/2$ (la mitad del arreglo), manteniendo una complejidad lineal $O(n)$ pero con un factor constante menor en el uso real.

1. Implementa una versión recursiva para el cálculo del factorial ($n! = n \cdot (n-1)!$) y una versión iterativa para Fibonacci.

````python
#factorial recursivo ( n!=n⋅(n−1)! )
def factorial_recursivo(n: int) -> int:
    if n < 0:
        raise ValueError("El factorial no está definido para números negativos.")
    if n in (0, 1):
        return 1
    return n * factorial_recursivo(n - 1)
````

````python
#Fibonacci iterativo
def fibonacci_iterativo(n: int) -> int:
    if n < 0:
        raise ValueError("El índice no puede ser negativo.")
    if n == 0:
        return 0

    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b
````
